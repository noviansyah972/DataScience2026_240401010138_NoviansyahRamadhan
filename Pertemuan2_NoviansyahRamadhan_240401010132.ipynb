````python
import pandas as pd
import numpy as np

url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df  = pd.read_csv(url)

df.head()
````

````python
# Q1 : Berapa jumlah baris dan kolom dataset Titanic
print (df.shape)
````

Untuk mencari jumlah baris dan kolom pada dataset bisa menggunakan fungsi df.shape pada library pandas.

Pada output df.shape (891, 12) menunjukan bahwa jumlah baris dan kolom pada data csv adalah 891 baris dan 12 kolom

````python
# Q2 Kolom mana yang memiliki nilai kosong (missing)?
print(df.isnull().sum())
````

Untuk mencari kolom yang memiliki nilai kosong (missing) bisa menggunakan fungsi df.isnull dan dikombinasikan dengan operasi sum untuk menjumlahkan data yang kosong pada setiap kolom.

Output yang di hasilkan dari kombinasi (df.isnull().sum()) akan menampilkan jumlah setiap data kosong (missing/null) pada kolom.



````python
# Q3 Berapa persen penumpang yang selamat (Survived=1)?
survived_passenger = df['Survived'].sum()
survived_passenger_precentage  = df['Survived'].mean()*100

print(f"Penumpang yang selamat adalah {survived_passenger}")
print(f"Persentase penumpang yang selamat adalah {survived_passenger_precentage:.5f}%")
````

Untuk mencari persentase penumpang yang selamat, bisa menjumlahkan dahulu kolom survived dengan fungsi

`df['Survived'].sum()`


lalu bisa melajutkan mencari rata rata (mean) antara data survived dengan seluruh data dan di kalikan 100 untuk mengubah ke persen

`df['Survived'].mean()*100`


setelah melakukan program tersebut, maka output bisa menggunakan variable yang sudah di buat

`print(f"Penumpang yang selamat adalah {survived_passenger}")`

`print(f"Persentase penumpang yang selamat adalah {survived_passenger_precentage:.5f}%")`


````python
# Q4 Rata-rata usia penumpang wanita kelas 1 yang selamat?
wk1 = df[
  (df['Sex']=='female') & #Seleksi gender
  (df['Pclass']==1) &     #Seleksi kelas
  (df['Survived']==1)     #seleksi selamat/tidak
]
print (f"Rata-rata usia penumpang wanita kelas 1 yang selamat:{wk1['Age'].mean():.2f} tahun")
````

Lakukan seleksi data terlebih dahulu menggunakan boolean indexing sebagai berikut

```

  (df['Sex']=='female') & #Seleksi gender
  (df['Pclass']==1) &     #Seleksi kelas
  (df['Survived']==1)     #seleksi selamat/tidak
]
```
Maka bisa di lanjutkan mencari rata-rata usia wanita kelas 1 yang selamat dan menjadikannya output

`print (f"Rata-rata usia penumpang wanita kelas 1 yang selamat:{wk1['Age'].mean():.2f} tahun")`


````python
# Q5 Kelas penumpang mana yang tingkat keselamatannya tinggi?
hasil = df.groupby('Pclass').agg(
    total=('PassengerId', 'count'),         #jumlah penumpang berdasarkan kelas
    selamat=('Survived','sum'),             #jumlah penumpang selamat berdasarkan kelas
    tingkat_keselamatan=('Survived','mean') #perhitungan rata rata penumpang selamat dari jumlah penumpang
)
hasil['tingkat_keselamatan'] = hasil['tingkat_keselamatan'] * 100

print(hasil.round(2))
````

Mencari kelas yang memiliki tingkat keselamatan paling tinggi bisa menggunakan `df.groupby` untuk mengelompokkan data berdasarkan suatu kolom. pada case ini pengelompokan di lakukan pada kolom class,

dilanjutkan menggunakan aggregate untuk melakukan beberapa perhitungan sekaligus pada setiap kelompok.

````python
# Q6 Gender penumpang yang memiliki tingkat keselamatan tinggi?
hasil = df.groupby('Sex').agg(
    total=('Sex', 'count'),         #jumlah penumpang berdasarkan gender
    selamat=('Survived','sum'),             #jumlah penumpang selamat kelas x
    tingkat_keselamatan=('Survived','mean') #perhitungan rata rata penumpang selamat dari jumlah penumpang berdasarkan gender
)
hasil['tingkat_keselamatan'] = hasil['tingkat_keselamatan'] * 100

print(hasil.round(2))
````
